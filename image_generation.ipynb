```python
from PIL import Image, ImageDraw, ImageFont
import os

out_dir = r'c:\wdd231\chamber\images'
os.makedirs(out_dir, exist_ok=True)
labels = ['Heritage','Landmark','Culture','Adventure','Scenery','Crafts','Events','Experience']
colors = [(64,132,196),(187,89,58),(70,167,135),(212,145,54),(129,89,179),(35,120,104),(185,52,86),(41,98,151)]
for i, (color, label) in enumerate(zip(colors, labels), start=1):
    img = Image.new('RGB', (640, 420), color)
    draw = ImageDraw.Draw(img)
    draw.rectangle((40, 40, 600, 380), fill=(255,255,255,180))
    draw.rectangle((40, 40, 600, 110), fill=(18, 23, 42))
    try:
        font = ImageFont.truetype('arial.ttf', 28)
        font2 = ImageFont.truetype('arial.ttf', 18)
    except OSError:
        font = ImageFont.load_default()
        font2 = ImageFont.load_default()
    draw.text((70, 65), f'Timbuktu #{i}', fill=(255,255,255), font=font)
    draw.text((70, 130), label, fill=(18, 23, 42), font=font)
    draw.text((70, 190), 'Discover Chamber', fill=(18, 23, 42), font=font2)
    img.save(os.path.join(out_dir, f'discover-{i}.webp'), 'WEBP', quality=90)
print('created', i, 'files')
```
